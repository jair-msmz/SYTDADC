## Vectorizar el texto usando TF-IDF

```python
vectorizador = TfidfVectorizer(stop_words='english')
```

## Modelo de clasificación simple

```python
modelo_clasificacion = make_pipeline(vectorizador, MultinomialNB())
```

```python
modelo_clasificacion.fit(df_documentos['texto'], df_documentos['categoria'])
print("Modelo entrenado exitosamente con TF-IDF y Naive Bayes.\n")
```
